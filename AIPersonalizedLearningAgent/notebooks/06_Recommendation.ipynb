# 06. Learning Resource Recommendation Model (Model 5)

This notebook fits an Item-Based Collaborative Filtering engine on OULA student interaction click logs, computes cosine similarity matrices, and evaluates Top-K resource recommendations using Precision@K, Recall@K, NDCG@K, and Hit Rate@K.

```python
import sys
import json
from pathlib import Path
cwd = Path.cwd()
workspace_dir = cwd.parents[1] if cwd.name == 'notebooks' else (cwd.parent if cwd.name == 'PersonalizedLearningAgent' else cwd)
sys.path.append(str(workspace_dir))

models_dir = workspace_dir / "PersonalizedLearningAgent" / "models"

with open(models_dir / "metrics_report.json", "r") as f:
    metrics = json.load(f)
    
print("Item-Based Collaborative Filtering Recommendation Metrics:")
for k, summary in metrics.items():
    print(f"  K = {k}:")
    for m, val in summary.items():
        print(f"    {m:<12}: {val:.4f}")
```
