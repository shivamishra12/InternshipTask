```python
from sklearn.preprocessing import LabelEncoder

oe = OrdinalEncoder()
df_encoded = df.copy()
categorical_cols = ['Building Type', 'Day of Week']
df_encoded[categorical_cols] = oe.fit_transform(df[categorical_cols])
df_encoded.head()
```

```python
corr_matrix = df_encoded.corr()

# Wrap 'Energy Consumption' in quotes
energy_correlation = corr_matrix['Energy Consumption'].sort_values(ascending=False)

print("Correlation of features relative to Energy Consumption:")
print(energy_correlation)
```

```python
from sklearn.linear_model import LinearRegression
```

```python
df_encoded.head()
```

```python
x_train = df_encoded.iloc[:, :-1]
y_train= df_encoded['Energy Consumption']
x_train
```

```python
y_train
```

```python
model = LinearRegression()
model
```

```python
model.fit(x_train,y_train)
```

```python
df_test = pd.read_csv('/kaggle/input/datasets/govindaramsriram/energy-consumption-dataset-linear-regression/train_energy_data.csv')
```

```python
df_test_encoded = df_test.copy()
df_test_encoded[categorical_cols] = oe.transform(df_test[categorical_cols])
df_test_encoded.head()
```

```python
from sklearn.metrics import r2_score, mean_squared_error, mean_absolute_error
X_test = df_test_encoded.drop(columns=['Energy Consumption'])
y_test = df_test_encoded['Energy Consumption']
y_pred = model.predict(X_test)
r2 = r2_score(y_test, y_pred)
mse = mean_squared_error(y_test, y_pred)
rmse = np.sqrt(mse)  # Root Mean Squared Error
mae = mean_absolute_error(y_test, y_pred)
```

```python
print("--- Model Performance Metrics ---")
print(f"R-squared Score (Variance Explained): {r2:.4f}")
print(f"R-squared Percentage Accuracy:       {r2 * 100:.2f}%")
print(f"Mean Absolute Error (MAE):           {mae:.2f} kWh")
print(f"Root Mean Squared Error (RMSE):      {rmse:.2f} kWh")
```
