# 04. EdNet Data Generation & Preprocessing (Model 3 Prep)

This notebook simulates student-question interactions using Item Response Theory (Rasch model) and cleans the dataset to build sliding sequences of length 50 for Deep Knowledge Tracing.

```python
import sys
from pathlib import Path
cwd = Path.cwd()
workspace_dir = cwd.parents[1] if cwd.name == 'notebooks' else (cwd.parent if cwd.name == 'PersonalizedLearningAgent' else cwd)
sys.path.append(str(workspace_dir))

import pandas as pd
from PersonalizedLearningAgent.ml.dkt.preprocess import preprocess_ednet_data
from PersonalizedLearningAgent.ml.dkt.dataset import get_dataloaders

raw_dir = workspace_dir / "PersonalizedLearningAgent" / "data" / "raw" / "ednet"
processed_dir = workspace_dir / "PersonalizedLearningAgent" / "data" / "processed"

# Clean and prepare EdNet
df_prep = preprocess_ednet_data(raw_dir / "ednet_data.csv", processed_dir / "ednet_preprocessed.csv")
print("Preprocessed Interactions Shape:", df_prep.shape)

# Extract dataloaders to verify sequence generation
train_loader, val_loader, test_loader = get_dataloaders(processed_dir / "ednet_preprocessed.csv", num_questions=500)
print(f"Sequence Dataloaders - Train: {len(train_loader.dataset)} | Val: {len(val_loader.dataset)} | Test: {len(test_loader.dataset)}")
```
