```python
from qiskit import QuantumCircuit
# Initialize one Qubit
qc=QuantumCircuit(1)
# Apply H gate
ans = qc.h(0)
# Drawing the qubit
print(qc.draw())
```

```python
from qiskit.quantum_info import Statevector
st =  Statevector.from_instruction(qc)
counts = st.sample_counts(shots=1000)

print("\n--- Measurement Results ---")
print(counts)
```
