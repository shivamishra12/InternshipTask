# 05. Deep Knowledge Tracing LSTM Model (Model 3)

This notebook trains and evaluates the Deep Knowledge Tracing LSTM model, computing ROC-AUC performance metrics and saving the trained neural network state-dict.

```python
import sys
import json
from pathlib import Path
cwd = Path.cwd()
workspace_dir = cwd.parents[1] if cwd.name == 'notebooks' else (cwd.parent if cwd.name == 'PersonalizedLearningAgent' else cwd)
sys.path.append(str(workspace_dir))

models_dir = workspace_dir / "PersonalizedLearningAgent" / "models"

with open(models_dir / "dkt_metrics.json", "r") as f:
    metrics = json.load(f)
    
print("Deep Knowledge Tracing LSTM Test Metrics:")
for m, val in metrics.items():
    print(f"  {m:<15}: {val}")
```
